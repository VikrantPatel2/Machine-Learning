### Magic Methods

Magic methods in Python, also known as dunder methods(double underscore methods), are special methods that start and end with double underscores.These methods enable you to define the behavior of objecrs for built-in operations, such as airtmetic operations, comparisons,and more.

## Magic Methods


Magic methods are predefined methods in Python that you can override to change the behavior of your objects.

```python
class Person:
        def  __init__(self,name,age):
            self.name=name
            self.age=age
        def __str__(self):
              return f"{self.name},{self.age} years old "
        
person=Person('Jack',21)
print(person)
```
