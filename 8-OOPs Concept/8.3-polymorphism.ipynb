### Polymorphism

It is a core concept of OOP that allow objects pf different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. Polymorphism is typically achieved through method overriding and interfaces.

### Method Overriding

Mwthod overriding allows a child class to provide a specific implementation of a method that is already defined in its parent class.

```python
### Create base class
class Animal:
    def speak(self):
        return "Sound of the animal"

## Derived class 1
class Dog(Animal):
    def speak(self):
        return 'Woof !'
    
 ## Derived calss
class Cat(Animal):   
    def speak(self):
        return "Meow !"

## Function that demonstrates Polymorphism    
def animal_speak(animal):
    print(animal.speak())
    
dog=Dog()
cat=Cat()
print(dog.speak())    
print(cat.speak())
animal_speak(dog)
```

```python
### Polymorphism with Functions and Methods
## base class

class Shape:
    def are(self):
        return "The area of the figure"

## Derived class
class Rectangle(Shape):
       def __init__(self,length,width):
            self.width=width
            self.length=length
           
       def  area(self):
            return self.width*self.length

class Circle(Shape):
     def __init__(self,radius):
          self.radius=radius

     def  area(self):
          return 3.14*self.radius**2               
## Function that demonstrates polymorphism

def print_area(shape):
     print(f"The area is {shape.area()}")

rectangle=Rectangle(4,5)
circle=Circle(3)

print_area(rectangle)
print_area(circle)
```

### Abstract Base Classes(ABCs)
These are used todefine common mrethods for a group of related objects.
They can enforce that derived classes implement particular methods,promoting consistency across different implementations.

```python
## Polymorphism with Abstarct Base Classes

from abc import ABC,abstractmethod

##Define an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

## Derived Class   1

class Car(Vehicle):
    def  start_engine(self):
        return "Car engine Started"
    
## Derived Class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motocycle Engine started"
    
## Function that demonstrates polymorphism
def start_vehicle(vehicle):
    print(vehicle.start_engine())
## created objects of car and motorcycle    
car=Car()
motocycle=Motorcycle()

start_vehicle(car)

```
