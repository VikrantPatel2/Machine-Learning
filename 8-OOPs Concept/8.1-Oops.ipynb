```python
## Instance Variable and Method
class Dog:
    ## Constructor
    def __init__( self,name,age):
        self.name=name
        self.age=age
##Onject
dog1=Dog('Buddy',3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2=Dog('Luci',4)
print(dog2.name)
print(dog2.age)
```

```python
### Define class with instance method
class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def Bark(self):
        print(f"{self.name} says woof")
dog1=Dog("Buddy",3)                    
dog1.Bark()
```

```python
## Modeling a Bank Account

## Define a class for bank account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. Now balance is {self.balance}") 

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient Balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. New Balance is{self.balance}") 

    def get_balance(self):
        return self.balance                  
    
### Create  an  account

account=BankAccount('Jack',2500*2)   
print(account.balance)
```

```python
### Callinstance methods
account.deposit(500)
```

```python
account.withdraw(100)
```

```python
print(account.get_balance())
```
