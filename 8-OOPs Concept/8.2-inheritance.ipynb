###  Inheritance In Python
Inheritance is a fundamental concept is OOP that allowsa class to inherit attributes and method from another class.

```python
## Inheritance

## Parent class
class Car:
    def __init__(self,windows,door,enginetype):
        self.windows=windows
        self.door=door
        self.enginetype=enginetype
    def drive(self):
            print(f"The person will drive the {self.enginetype} car")



```

```python
car1=Car(4,5,'Petrol')
print(car1.drive())
```

```python
class Tesla(Car):
    def __init__(self, windows, door, enginetype,is_selfdriving):
        super().__init__(windows, door, enginetype)
        self.is_selfdriving=is_selfdriving
    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_selfdriving}")    
```

```python
tesla1=Tesla(4,5,'Electric',True)
print(tesla1.selfdriving())
```

```python
tesla1.drive()
```

```python
## Multiple Inheritance
 ## WHen a class inherits from more than one class.
## Base class 1
class Animal:
    def __init__(self,name):
        self.name=name
    
    def speak(self):
        print("Subclass must implement this method")

## Base class 2
class Pet:
    def __init__(self,owner):
        self.owner=owner       

## Derive class
class Dog(Animal,Pet):
    def __init__(self, name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)     

    def speak(self):
        return f"{self.name} saywoof"
    
## Create an object
dog=Dog("Buddy","Jack")
print(dog.speak())    

print(f"Owner:{dog.owner}")
 
```
