```python
## for loop

for i in range(5):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,0,-1):
    print(i)
```

```python
## stringd
str="Vikrant Patel"
for i  in str:
    print(i)
```

```python
## while loop

count=0
while count<=5:
    print(count)
    count=count+1 
```

```python
## loop contro statement
## break- it exits the loop prematurely

for i in range(10):
    if(i==5):
        break
    print(i)

```

```python
## continue -- this statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)


```

```python
## pass- this statement is a null operation ; it does nothing

for i  in range(5):
    if i==3:
        print("The number is",i)
        pass
    print(i)
```

```python
## Nested loops
for i in range(3):
    for j in range(2):
        print(f'i:{i} and j:{j}')


```
