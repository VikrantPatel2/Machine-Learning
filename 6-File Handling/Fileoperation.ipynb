### File Operation- Read and Write Files

File handling is a crucial part of any programming language. Python provides built-in  functions and methods to read from and to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.

```python
## Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
### Read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() remove the newline character
```

```python
## Writing a file(Overwriting)
with open('example.txt','w') as file:
    file.write('Hello World!\n')
    file.write('This is a new line.')
```

```python
## write a file without overwriting

with open('example.txt','a') as file:
    file.write('Append operation taking place!\n')
```

```python
## writing list of lines to a file
lines=['First line\n','Second Line\n','THird Line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary Files
## writing to a binary file

data=b'\x00\x01\x02\x04'
with open('example.bin','wb')as file:
    file.write(data)
```

```python
## Reading a binary file
with open('example.bin','rb')as file:
    content=file.read()
    print(content)
```

```python
## Read the content from a source text files and write to a destination text file

with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w')as destination_file:
    destination_file.write(content)    
```

```python
with open('example.txt','w+') as file:
    file.write('Hello World\n')
    file.write('This is a new line\n')

    # Move cursor to beginning
    file.seek(0)

    # Read content
    content = file.read()
    print(content)
```
