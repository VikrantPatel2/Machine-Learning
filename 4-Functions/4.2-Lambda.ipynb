Lambda functions are small anonymous defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

```python
# syntax
lambda arguments:expression
```

```python
addition=lambda a,b:a+b
print(addition(2,3))
type(addition)
```

```python
even=lambda num:num%2==0
even(13)
```

```python
add1=lambda a,b,c:a+b+c
print(add1(1,23,4))
```

```python
## map()- applies a function to all items in a list

lst=[1,2,3,4,5,6]
square_num=list(map(lambda x :x**2, lst))
print(square_num)
```
