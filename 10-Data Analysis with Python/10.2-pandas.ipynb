```python
import pandas as pd
```

```python
### Series -- it is a one dimensional array like object that can hold any type data.

data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n", series)
```

```python
 ### Create a Series from dictonary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
print(type(series_dict))
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
### Data frame-

## create a Dataframe from a dictonary of list

data={
    'Name':['Jack','John','Arther'],
    'Age':[21,25,35],
    'City':['LA','Paris','Lucknow']
}

df=pd.DataFrame(data)
print(df)
```

```python
### Create a Data frame from a list of Dictonaries
data=[
    {'Name':'Jack','age':21,'City':'Lucknow'},
     {'Name':'John','age':21,'City':'Hardoi'},
      {'Name':'Jaugar','age':21,'City':'Mallawan'},
       {'Name':'Joker','age':21,'City':'Bilgram'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
