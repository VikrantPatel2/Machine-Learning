```python
###  Label Encoding involves assigning a unique label to each category in the variable. The labels are usually assigned in alphabetical order or based on the frequency of the categories.
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
## Create a simple dataframe
df=pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df.head()
```

```python
lbl_encoder=LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
#### Ordinal Encoding
### It is used to encode categrical data that have an intrinsic order or ranking,.Each category is assigned a numerical value based on its position in the order
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df.head()
```

```python
### create an instances of ordinaland then fit_transform
encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

```python
### Target Guided Ordinal Encoding
### It is a technique to encode categorical variables based on their releationship with the target variable. This encoding technique is useful when we have a categorical variable with a large number of unique categories.
```

```python
df=pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,180,320]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
df[['city','city_encoded']]
```
