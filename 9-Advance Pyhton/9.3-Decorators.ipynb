### Decorators

These are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionaly to functions or methods without modifying their actual code.

```python
## function copy
## closure
### decorators

def welcome():
    return "Welcome to course"

welcome()
```

```python
wel=welcome
wel()
```

```python
del welcome
```

```python
wel()
```

```python
## Closures

def main_welcome():
    msg='Welcome'
    def sub_welcome_method():
        print("Welcome to the advance pythom course")
        print(msg)
        print("Please learn the concept")

    return sub_welcome_method()    
```

```python
main_welcome()
```

```python

def main_welcome(func):
    msg='Welcome'
    def sub_welcome_method():
        print("Welcome to the advance pythom course")
        func()
        print("Please learn the concept")

    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python

def main_welcome(func,lst):
   
    def sub_welcome_method():
        print("Welcome to the advance pythom course")
        print(func(lst))
        print("Please learn the concept")

    return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5])
```

```python
## Decorator

def main_welcome(func):
    
    def sub_welcome_method():
        print("Welcome to the advance pythom course")
        func()
        print("Please learn the concept")

    return sub_welcome_method()

@main_welcome
def course_introduction():
    print("This is an advanced python course")
```
