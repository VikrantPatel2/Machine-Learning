Dictonaries are unordered collections of items. They store data in key-values pairs. Keys must be unique and immutable(e.g., strings,numbers, or tuples), while values can be of any type.

```python
## creating Dictonaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={'name':'Vikrant','age':20,'grade':24}
print(student)
print(type(student))
```

```python
## Accessing Dictonary Elements
student={'name':'Vikrant','age':20,'grade':'A'}
print(student['grade'])
print(student['age'])

print(student.get('grade'))
```

```python
##Modifying Dictonary Elements
## Dictonary are mutable
print(student)
```

```python
student['age']=33
student['address']='Lucknow'
print(student)
student['grade']='A'
del student['grade'] ## delete key and value pair
print(student)
```

```python
## Dictonary Methods

keys=student.keys()
print(keys)
values=student.values()
print(values)

items=student.items()
print(items)
```

```python
## shallow copy
student_copy=student
print(student_copy)
```

```python
student['name']='Vikrant1'
print(student)
print(student_copy)
```

```python
## shallow copy
student_copy1=student.copy()
print(student_copy1)
```

```python
## Iterating Over Dictonaries
## You can use loops to iterative over dictonaries

##iterating over keys
for key in student.keys():
    print(key)
```

```python
## Iterative over values
for value in student.values():
    print(value)
```

```python
for  key,value in student.items():
    print(f'{key}:{value}')
```

```python
## Nested Dictonaries

students={
    'student1':{'name':'Vikrant','age':21},
    'student2':{'name':'Jack','age':21},
    'student3':{'name':'Daniel','age':26}
}
print(students)
```

```python
## Accesss nested dictonaries elements
print(students['student2']['name'])
print(students['student1']['age'])
```

```python
## Iterating over nested dictonaries

for student_id, student_info in students.items():
    print(f'{student_id}:{student_info}')
    for key, value in student_info.items():
        print(f'{key}:{value}')
```

```python
## Dictonary Comprehension

square={x:x**2 for x in range(5)}
print(square)
```

```python
## Condition dictonary Comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Examples

## use a dictonary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for num in numbers:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)            
```

```python
## Merge two dictonaries into one
dict1={'a':1,'b':2}
dict2={'b':3,'c':4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
