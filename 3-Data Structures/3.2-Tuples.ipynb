Tuples are ordered collections of items that
 are immutable . They are similar to lists but their immutability makes them different.

```python
## Tuple Methods

print(numbers.count(1))
print(numbers.index(3))
```

```python
## packing and unpacking of tuple
packed_tuple=1,'Hello',3.14
print(packed_tuple)
```

```python
##Unpacking tuple

a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python

## Nested List
lst=[[1,2,3],[6,8,9,7],["hello",1,3.14,'c']]
lst[0][0:3]

```

```python
## Nested Tuples
nested_tuple=((1,2,3),('a','b','c'),(True,False))

print(nested_tuple[0])
print(nested_tuple[1][2])
```
